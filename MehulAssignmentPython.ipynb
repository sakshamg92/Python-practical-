```python
n = input("Enter a string:")
vowel=0
consonant=0
digit = 0
spaces = 0
special = 0
for i in n:
  if i.lower() in 'aeiou':
    vowel +=1
  elif i.isalpha():
    consonant +=1
  elif i.isdigit():
    digit +=1
  elif i.isspace():
    spaces +=1
  else:
    special +=1
print("Vowels:", vowel)
print("Consonants:", consonant)
print("Digits:", digit)
print("Spaces:", spaces)
print("Special characters:", special)
```

```python
class Student:
    def __init__(self, name, roll_no, marks):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks

    def display_details(self):
        print("--- Student Details ---")
        print("Name:    ", self.name)
        print("Roll No: ", self.roll_no)
        print("Marks:   ", self.marks)

    def calculate_grade(self):
        if self.marks >= 90 and self.marks <= 100:
            print("Grade:    A")
        elif self.marks >= 80 and self.marks < 90:
            print("Grade:    B")
        elif self.marks >= 70 and self.marks < 80:
            print("Grade:    C")
        elif self.marks >= 60 and self.marks < 70:
            print("Grade:    D")
        else:
            print("Grade:    Fail")


stu = Student("Maverick", 17, 95)
stu.display_details()
stu.calculate_grade()
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display_details(self):
        print(f"Employee Name: {self.name}")
        print(f"Salary: ${self.salary}")
        print("Role: General Employee")

class Manager(Employee):
    def __init__(self, name, salary, department):
        super().__init__(name, salary)
        self.department = department


    def display_details(self):
        print(f"Employee Name: {self.name}")
        print(f"Salary: ${self.salary}")
        print(f"Role: Manager of {self.department} Department")

class Developer(Employee):
    def __init__(self, name, salary, programming_language):
        super().__init__(name, salary)
        self.programming_language = programming_language

    def display_details(self):
        print(f"Employee Name: {self.name}")
        print(f"Salary: ${self.salary}")
        print(f"Role: Developer specialized in {self.programming_language}")

print("--- Developer Details ---")
dev = Developer("Alice", 95000, "Python")
dev.display_details()

print("\n--- Manager Details ---")
man = Manager("Bob", 120000, "Engineering")
man.display_details()

print("\n--- Generic Employee Details ---")
emp = Employee("Charlie", 60000)
emp.display_details()
```

```python
import time


def measure_time(func):
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result = func(*args, **kwargs)
        end_time = time.time()
        execution_time = end_time - start_time
        print(f"Execution time of '{func.__name__}': {execution_time:.6f} seconds")
        return result
    return wrapper


@measure_time
def calculate_sum(n):
    total_sum = 0
    for i in range(1, n + 1):
        total_sum += i
    return total_sum


limit = 100000
result = calculate_sum(limit)
print(f"Sum of numbers from 1 to {limit:,}: {result}")
```

```python
def fibonacci_generator(limit):
    a, b = 0, 1
    while a <= limit:
        yield a
        a, b = b, a + b


limit_val = 100
print(f"Fibonacci numbers up to {limit_val}:")
for num in fibonacci_generator(limit_val):
    print(num, end=" ")
print()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

data = {
    'Student': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Math': [85, 92, 78, 89, 95],
    'Science': [90, 88, 85, 94, 91],
    'English': [78, 85, 82, 80, 88]
}
df = pd.DataFrame(data)
df.set_index('Student', inplace=True)


fig, ax = plt.subplots(figsize=(8, 5))
x = np.arange(len(df.index))
width = 0.25


ax.bar(x - width, df['Math'], width, label='Math', color='#1f77b4')
ax.bar(x, df['Science'], width, label='Science', color='#ff7f0e')
ax.bar(x + width, df['English'], width, label='English', color='#2ca02c')


ax.set_ylabel('Marks')
ax.set_title('Marks Comparison by Student and Subject')
ax.set_xticks(x)
ax.set_xticklabels(df.index)
ax.legend()
ax.set_ylim(0, 110)

plt.tight_layout()
plt.show()


plt.figure(figsize=(7, 4))
sns.heatmap(df, annot=True, cmap='coolwarm', fmt='d', cbar_kws={'label': 'Marks'})
plt.title('Heatmap of Student Marks')
plt.ylabel('Student')
plt.xlabel('Subject')
plt.tight_layout()
plt.show()
```

```python
import requests
from urllib.parse import urlencode


url = "https://httpbin.org/get"
params = {'name': 'Maverick'}

try:
    print("--- Sending GET request using requests library ---")

    response = requests.get(url, params=params, timeout=5)


    print(f"Status Code: {response.status_code}")
    print("JSON Response:")
    print(response.json())

except requests.exceptions.Timeout:
    print("Error: The request timed out!")
except requests.exceptions.ConnectionError:
    print("Error: A connection error occurred!")
except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")

print("\n--- Exploring urllib.parse.urlencode() ---")
query_params = {
    'name': 'Maverick Vinales',
    'course': 'Computer Science',
    'city': 'New York'
}

encoded_string = urlencode(query_params)
print("Original Dictionary:", query_params)
print("Encoded Query String:", encoded_string)
print("Full Constructed URL:", f"{url}?{encoded_string}")
```
